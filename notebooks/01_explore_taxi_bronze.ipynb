```python
df = spark.read.parquet("Files/bronze/yellow_tripdata_2025-06.parquet")
```

```python
df.printSchema()
```

```python
df.show(5, truncate=False)
```

```python
df_filtered1 = df.filter(df.tpep_pickup_datetime == df.tpep_dropoff_datetime)

df_filtered1.show(5)
```

```python
suspicious_trips = df_filtered1.count()
```

```python
total_trips = df.count()
```

```python
anomaly_percentage = suspicious_trips/total_trips*100
```

```python
print("Total trips: ", total_trips)
print("Anomaly percentage: ", round(anomaly_percentage, 2), "%")
```

```python
negative_fares = df.filter(df.fare_amount < 0).count()
```

```python
missing_passengers = df.filter(df.passenger_count.isNull()).count()
```

```python
print("Missing passengers:", missing_passengers)
print("Negative fares:", negative_fares)
```
